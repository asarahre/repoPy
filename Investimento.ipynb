```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

tickers = ['PETR4.SA', 'BBAS3.SA']

# Baixar os dados com auto_adjust=True (padrão atual)
data = yf.download(tickers, start='2023-05-01', end='2025-05-01', auto_adjust=True)['Close']

# Calcular retorno diário
daily_returns = data.pct_change().dropna()

# Calcular retorno cumulativo
cumulative_returns = (1 + daily_returns).cumprod() - 1

# Retorno médio diário
mean_daily_return = daily_returns.mean()

# Desvio padrão dos retornos diários
std_daily_return = daily_returns.std()

print("Retorno médio diário:")
print(mean_daily_return)
print("\nDesvio padrão dos retornos diários:")
print(std_daily_return)

# Gráfico de retorno cumulativo
plt.figure(figsize=(12, 6))
plt.plot(cumulative_returns.index, cumulative_returns['PETR4.SA'], label='PETR4')
plt.plot(cumulative_returns.index, cumulative_returns['BBAS3.SA'], label='BBAS3')

plt.title('Retorno Cumulativo Diário - Google vs Microsoft')
plt.xlabel('Data')
plt.ylabel('Retorno Cumulativo')
plt.legend()
plt.grid(True)
plt.show()
```
