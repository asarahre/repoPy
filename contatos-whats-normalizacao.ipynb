```python
import re

with open("/content/sample_data/contrato.txt", "r", encoding="utf-8") as arquivo:
    texto = arquivo.read()

contratos = re.findall(
    r"contrato\s*(?:nº|n°|:)?\s*(\d+)",
    texto,
    flags=re.IGNORECASE
)

# remove duplicados mantendo ordem
contratos = list(dict.fromkeys(contratos))

resultado = ",".join(f"'{c}'" for c in contratos)

with open("/content/contratos3.txt", "w", encoding="utf-8") as saida:
    saida.write(resultado)

print("Arquivo gerado com sucesso!")
print(resultado)
print(len(contratos))
```
